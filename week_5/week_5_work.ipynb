```python
import pandas as pd

# Load data
df = pd.read_excel("HousePrice.xlsx")

# Check structure
print(df.head())
print(df.columns)

# ---------- X (features) ----------
X = df.drop(columns=["days_on_market"]).copy()

# Convert categorical variables
X = pd.get_dummies(X)

# ---------- y (target) ----------
# Low = 1 (positive class), else 0
y = df["days_on_market"].apply(lambda x: 1 if x == "Low" else 0)

# Check results
print("\nX shape:", X.shape)
print("\ny distribution:")
print(y.value_counts())
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_validate

# Feature subset
X1 = X[[
    "number_of_beds",
    "number_of_baths",
    "number_of_parks",
    "property_size"
]]

# Model_1
Model_1 = DecisionTreeClassifier(max_depth=14, random_state=30)

# Cross-validation
cv_results_1 = cross_validate(
    Model_1,
    X1,
    y,
    cv=5,
    scoring=["accuracy", "recall", "precision", "f1"]
)

# Convert to DataFrame
cv_df_1 = pd.DataFrame(cv_results_1)

print(cv_df_1)

means = {
    "Accuracy": cv_df_1["test_accuracy"].mean(),
    "Recall": cv_df_1["test_recall"].mean(),
    "Precision": cv_df_1["test_precision"].mean(),
    "F1": cv_df_1["test_f1"].mean()
}

print(means)

# Get highest metric
print("\nHighest:", max(means, key=means.get))

# Averages
print("\nModel_1 Average Metrics")
print("Accuracy :", cv_df_1["test_accuracy"].mean())
print("Recall   :", cv_df_1["test_recall"].mean())
print("Precision:", cv_df_1["test_precision"].mean())
print("F1 Score :", cv_df_1["test_f1"].mean())

cv_df_1[["test_accuracy","test_recall","test_precision","test_f1"]].mean().sort_values(ascending=False)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Model_2
Model_2 = RandomForestClassifier(
    max_depth=14,
    n_estimators=12,
    random_state=0
)

# Cross-validation
cv_results_2 = cross_validate(
    Model_2,
    X1,
    y,
    cv=5,
    scoring=["accuracy", "recall", "precision", "f1"]
)

# Convert to DataFrame
cv_df_2 = pd.DataFrame(cv_results_2)

print(cv_df_2)

# Averages
print("\nModel_2 Average Metrics")
print("Accuracy :", cv_df_2["test_accuracy"].mean())
print("Recall   :", cv_df_2["test_recall"].mean())
print("Precision:", cv_df_2["test_precision"].mean())
print("F1 Score :", cv_df_2["test_f1"].mean())
```

```python
# Model_3
Model_3 = RandomForestClassifier(
    max_depth=14,
    n_estimators=12,
    random_state=0
)

# Cross-validation using ALL features
cv_results_3 = cross_validate(
    Model_3,
    X,
    y,
    cv=5,
    scoring=["accuracy", "recall", "precision", "f1"]
)

# Convert to DataFrame
cv_df_3 = pd.DataFrame(cv_results_3)

print(cv_df_3)

# Averages
print("\nModel_3 Average Metrics")
print("Accuracy :", cv_df_3["test_accuracy"].mean())
print("Recall   :", cv_df_3["test_recall"].mean())
print("Precision:", cv_df_3["test_precision"].mean())
print("F1 Score :", cv_df_3["test_f1"].mean())
```

```python
# Model_4
Model_4 = RandomForestClassifier(
    max_depth=28,
    n_estimators=24,
    random_state=0
)

# Cross-validation using ALL features
cv_results_4 = cross_validate(
    Model_4,
    X,
    y,
    cv=5,
    scoring=["accuracy", "recall", "precision", "f1"]
)

# Convert to DataFrame
cv_df_4 = pd.DataFrame(cv_results_4)

print(cv_df_4)

# Averages
print("\nModel_4 Average Metrics")
print("Accuracy :", cv_df_4["test_accuracy"].mean())
print("Recall   :", cv_df_4["test_recall"].mean())
print("Precision:", cv_df_4["test_precision"].mean())
print("F1 Score :", cv_df_4["test_f1"].mean())
```

```python
precision_scores = {
    "Model 1": cv_df_1["test_precision"].mean(),
    "Model 2": cv_df_2["test_precision"].mean(),
    "Model 3": cv_df_3["test_precision"].mean(),
    "Model 4": cv_df_4["test_precision"].mean()
}

print(precision_scores)

# Get best model
print("\nBest:", max(precision_scores, key=precision_scores.get))
```
