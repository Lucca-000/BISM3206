```python
import pandas as pd
import numpy as np
import sklearn as sk
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel('df.xlsx')

#df->df1a
df1a = df.copy()

#scaled
min_age = df1a["Age"].min()
max_age = df1a["Age"].max()

df1a["Scaled_Age"] = (df1a["Age"] - min_age) / (max_age - min_age)

#preview
df1a.head()
```

```python
#df1a
avg_age = df1a["Age"].mean()

#loop comparo
age_comp = []

for age in df1a["Age"]:
    if age > avg_age:
        age_comp.append("Higher")
    elif age < avg_age:
        age_comp.append("Lower")
    else:
        age_comp.append("Same")

df1a["Age_Comparison"] = age_comp

# preview
df1a[["Age", "Age_Comparison"]].head()
```

```python
df_quiz = df1a.loc[100:500, ["Age", "Scaled_Age", "Age_Comparison"]].reset_index(drop=True)
print(df_quiz)
```
