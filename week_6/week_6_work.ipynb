```python
# Imports
import pandas as pd
from sklearn.model_selection import cross_validate
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier

# Load data (UPDATE PATH)
df = pd.read_excel("BushfireData.xlsx")

# =========================
# Practice 1
# =========================
dfXX = df.copy()

# Drop columns with >3000 missing values
missing_counts = dfXX.isnull().sum()
cols_to_drop = missing_counts[missing_counts > 3000].index
dfXX = dfXX.drop(columns=cols_to_drop)

# Drop date column
if "date" in dfXX.columns:
    dfXX = dfXX.drop(columns=["date"])

# =========================
# Practice 2
# =========================
# Fill remaining missing values with column mean
for col in dfXX.columns:
    if dfXX[col].isnull().sum() > 0:
        dfXX[col] = dfXX[col].fillna(dfXX[col].mean())

# =========================
# Practice 3
# =========================
# Convert categorical → numerical
dfXX = pd.get_dummies(dfXX)

# =========================
# Practice 4
# =========================
dfXX = dfXX.iloc[0:10000]

# =========================
# Practice 5
# =========================
y = dfXX["ReductionBurn_Tomorrow"]
X = dfXX.drop(columns=["ReductionBurn_Tomorrow"])

# =========================
# Practice 6 - KNN
# =========================
My_KNN = KNeighborsClassifier(n_neighbors=5)

cv_knn = cross_validate(
    My_KNN,
    X,
    y,
    cv=5,
    scoring=["accuracy", "recall", "precision", "f1", "roc_auc"]
)

results_knn = {
    metric: cv_knn[f"test_{metric}"].mean()
    for metric in ["accuracy", "recall", "precision", "f1", "roc_auc"]
}

print("KNN Results:", results_knn)

# =========================
# Practice 7 - Naive Bayes
# =========================
My_NB = GaussianNB()

cv_nb = cross_validate(
    My_NB,
    X,
    y,
    cv=5,
    scoring=["accuracy", "recall", "precision", "f1", "roc_auc"]
)

results_nb = {
    metric: cv_nb[f"test_{metric}"].mean()
    for metric in ["accuracy", "recall", "precision", "f1", "roc_auc"]
}

print("NB Results:", results_nb)

# =========================
# Practice 8 - XGBoost
# =========================
My_XGB = XGBClassifier(
    max_depth=30,
    learning_rate=0.02,
    random_state=10,
    use_label_encoder=False,
    eval_metric="logloss"
)

cv_xgb = cross_validate(
    My_XGB,
    X,
    y,
    cv=5,
    scoring=["accuracy", "recall", "precision", "f1", "roc_auc"]
)

results_xgb = {
    metric: cv_xgb[f"test_{metric}"].mean()
    for metric in ["accuracy", "recall", "precision", "f1", "roc_auc"]
}

print("XGB Results:", results_xgb)

My_XGB = XGBClassifier(
    max_depth=30,
    learning_rate=0.02,
    random_state=10,
    eval_metric="logloss"
)
```
